```python
"""
graphvis_lp/scripts/analyze_weights.py

Анализ распределения semantic_weight на рёбрах.

Usage:
    python -m scripts.analyze_weights
    python -m scripts.analyze_weights --top 20
    python -m scripts.analyze_weights --export weights.csv
"""

from __future__ import annotations

import logging
import sys
from typing import Any, Dict, List, Optional

import numpy as np
from neo4j import GraphDatabase

from config import Neo4jConfig

logger = logging.getLogger(__name__)


class WeightAnalyzer:
    """Анализ распределения semantic_weight и transe_score на рёбрах."""

    def __init__(self, neo4j_cfg: Neo4jConfig):
        self._driver = GraphDatabase.driver(
            neo4j_cfg.uri, auth=(neo4j_cfg.user, neo4j_cfg.password),
        )
        self._db = neo4j_cfg.database

    def __enter__(self) -> "WeightAnalyzer":
        return self

    def __exit__(self, *exc) -> None:
        self._driver.close()

    def _session(self):
        return self._driver.session(database=self._db)

    def load_weights(
        self,
        weight_property: str = "transe_score",
        limit: Optional[int] = None,
    ) -> List[Dict[str, Any]]:
        """Загружает рёбра с весами."""
        query = (
            "MATCH (h:Entity)-[r:RELATION]->(t:Entity) "
            f"WHERE r.{weight_property} IS NOT NULL "
            "RETURN h.name AS head, r.type AS rel, t.name AS tail, "
            f"  r.{weight_property} AS weight"
        )
        if limit:
            query += f" LIMIT {int(limit)}"

        with self._session() as s:
            return [dict(r) for r in s.run(query)]

    def distribution_report(
        self,
        weight_property: str = "transe_score",
    ) -> Dict[str, Any]:
        """Полный статистический отчёт."""
        records = self.load_weights(weight_property)
        if not records:
            logger.error("Нет рёбер с %s", weight_property)
            return {}

        weights = np.array([r["weight"] for r in records], dtype=np.float64)
        n = len(weights)

        # Базовая статистика
        stats = {
            "count": n,
            "mean": float(np.mean(weights)),
            "std": float(np.std(weights)),
            "min": float(np.min(weights)),
            "max": float(np.max(weights)),
            "median": float(np.median(weights)),
        }

        # Перцентили
        percentiles = [1, 5, 10, 25, 50, 75, 90, 95, 99]
        stats["percentiles"] = {
            f"p{p}": float(np.percentile(weights, p))
            for p in percentiles
        }

        # Гистограмма (10 бинов)
        hist_counts, hist_edges = np.histogram(weights, bins=10)
        stats["histogram"] = [
            {
                "range": f"[{hist_edges[i]:.3f}, {hist_edges[i+1]:.3f})",
                "count": int(hist_counts[i]),
                "pct": f"{hist_counts[i] / n * 100:.1f}%",
            }
            for i in range(len(hist_counts))
        ]

        # Проверка на «подозрительные» распределения
        stats["warnings"] = []
        if stats["std"] < 0.05:
            stats["warnings"].append(
                f"ОЧЕНЬ НИЗКАЯ дисперсия (std={stats['std']:.4f}). "
                f"Все веса ~{stats['mean']:.3f}. Веса неинформативны!"
            )
        if stats["min"] == stats["max"]:
            stats["warnings"].append("ВСЕ ВЕСА ОДИНАКОВЫ. Что-то сломалось.")
        if stats["mean"] > 0.9:
            stats["warnings"].append(
                f"Среднее={stats['mean']:.3f}. Слишком высокое — "
                f"модель не различает далёкие/близкие пары."
            )
        if stats["mean"] < 0.1:
            stats["warnings"].append(
                f"Среднее={stats['mean']:.3f}. Слишком низкое — "
                f"эмбеддинги могут быть некорректными."
            )

        # Анализ по типам отношений
        rel_weights: Dict[str, List[float]] = {}
        for r in records:
            rel_weights.setdefault(r["rel"], []).append(r["weight"])

        stats["per_relation"] = sorted(
            [
                {
                    "relation": rel,
                    "count": len(ws),
                    "mean": float(np.mean(ws)),
                    "std": float(np.std(ws)),
                    "min": float(np.min(ws)),
                    "max": float(np.max(ws)),
                }
                for rel, ws in rel_weights.items()
            ],
            key=lambda x: x["mean"],
            reverse=True,
        )

        return stats

    def top_and_bottom(
        self,
        weight_property: str = "transe_score",
        k: int = 10,
    ) -> Dict[str, List[Dict]]:
        """Топ-K самых высоких и низких весов."""
        records = self.load_weights(weight_property)
        if not records:
            return {"top": [], "bottom": []}

        sorted_recs = sorted(records, key=lambda r: r["weight"], reverse=True)
        return {
            "top": sorted_recs[:k],
            "bottom": sorted_recs[-k:],
        }

    def print_report(
        self,
        weight_property: str = "transe_score",
        top_k: int = 10,
    ) -> None:
        """Печатает полный отчёт."""
        report = self.distribution_report(weight_property)
        if not report:
            return

        print(f"\n{'=' * 70}")
        print(f"  РАСПРЕДЕЛЕНИЕ {weight_property} ({report['count']:,} рёбер)")
        print(f"{'=' * 70}")

        print(f"\n  Базовая статистика:")
        print(f"    Mean   : {report['mean']:.4f}")
        print(f"    Std    : {report['std']:.4f}")
        print(f"    Min    : {report['min']:.4f}")
        print(f"    Max    : {report['max']:.4f}")
        print(f"    Median : {report['median']:.4f}")

        print(f"\n  Перцентили:")
        for name, val in report["percentiles"].items():
            print(f"    {name:>4s} : {val:.4f}")

        print(f"\n  Гистограмма:")
        max_count = max(h["count"] for h in report["histogram"])
        for h in report["histogram"]:
            bar_len = int(h["count"] / max_count * 30)
            bar = "█" * bar_len
            print(f"    {h['range']:>22s} | {bar:<30s} {h['count']:>7,} ({h['pct']})")

        if report["warnings"]:
            print(f"\n  ⚠ Предупреждения:")
            for w in report["warnings"]:
                print(f"    - {w}")

        # По типам отношений (топ-10 и bottom-10)
        per_rel = report["per_relation"]
        print(f"\n  Топ-5 отношений (высокий средний вес):")
        for r in per_rel[:5]:
            print(
                f"    {r['relation']:<25s} mean={r['mean']:.4f} "
                f"std={r['std']:.4f} n={r['count']:,}"
            )
        print(f"\n  Bottom-5 отношений (низкий средний вес):")
        for r in per_rel[-5:]:
            print(
                f"    {r['relation']:<25s} mean={r['mean']:.4f} "
                f"std={r['std']:.4f} n={r['count']:,}"
            )

        # Конкретные примеры
        tb = self.top_and_bottom(weight_property, top_k)
        print(f"\n  Топ-{top_k} самых высоких весов:")
        for r in tb["top"]:
            print(
                f"    ({r['head']}) -[{r['rel']}]-> ({r['tail']}) "
                f"= {r['weight']:.4f}"
            )
        print(f"\n  Топ-{top_k} самых низких весов:")
        for r in tb["bottom"]:
            print(
                f"    ({r['head']}) -[{r['rel']}]-> ({r['tail']}) "
                f"= {r['weight']:.4f}"
            )

        print(f"\n{'=' * 70}")


logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s [%(levelname)s] %(name)s: %(message)s",
)

top_k = 10
export_path = None
for i, arg in enumerate(sys.argv[1:], 1):
    if arg == "--top" and i < len(sys.argv) - 1:
        top_k = int(sys.argv[i + 1])
    if arg == "--export" and i < len(sys.argv) - 1:
        export_path = sys.argv[i + 1]

cfg = Neo4jConfig.from_env()

with WeightAnalyzer(cfg) as analyzer:
    # Отчёт по semantic_weight (cosine)
    analyzer.print_report("semantic_weight", top_k)

    # Отчёт по transe_score (если есть)
    analyzer.print_report("transe_score", top_k)

    # Экспорт
    if export_path:
        import csv
        records = analyzer.load_weights("transe_score")
        with open(export_path, "w", newline="", encoding="utf-8") as f:
            writer = csv.DictWriter(
                f, fieldnames=["head", "rel", "tail", "weight"],
            )
            writer.writeheader()
            writer.writerows(records)
        logger.info("Экспорт: %s (%d рёбер)", export_path, len(records))
```
